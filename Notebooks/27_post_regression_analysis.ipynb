**When to use it.** After regression, to turn coefficients into figures.

**What you get.** Ranked hits, volcano plots and per-gene detail.

## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.submodules.post_regression_analysis`](https://einarolafsson.github.io/spacr/api/spacr/submodules/index.html#spacr.submodules.post_regression_analysis)

- **`csv_file`** — CSV with columns ``grna``, ``fraction`` and ``prc``.
- **`grna_dict`** — mapping of anchor ``grna`` names to their fixed effect sizes.
- **`grna_list`** — gRNAs to include in the correlation matrix.
- **`save`** — persist correlation matrix, effect sizes and plots. Default ``False``.

## 4. Run

After editing the settings cell, run the function cell immediately below it. Long operations report progress through spaCR's logging system; set `SPACR_LOG_LEVEL=DEBUG` before starting Jupyter for more detail.

````python
settings = {
    'csv_file': None,
    'grna_dict': None,
    'grna_list': None,
    'save': False,
}
````

````python
post_regression_analysis(**settings)
````

Ranked hits, volcano plots and per-gene detail.
