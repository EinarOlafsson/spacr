## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.toxo.custom_volcano_plot`](https://einarolafsson.github.io/spacr/api/spacr/toxo/index.html#spacr.toxo.custom_volcano_plot)

- **`data_path`** — Regression table containing ``feature``, ``coefficient``, and ``p_value`` columns. DataFrame input is copied.
- **`draw`** — Build, optionally save, and show the figure. If false, return the hit list before constructing a figure.
- **`figsize`** — Width and height of the square figure in inches. Typography scales with this value.
- **`highlight_location`** — Values from ``metadata_column`` to overlay in the highlight colour and name in the in-panel legend.
- **`metadata_column`** — Localization or annotation column used by ``highlight_location``.
- **`metadata_path`** — Gene metadata containing one row per ``gene_nr`` and the selected metadata column. DataFrame input is copied.
- **`point_size`** — Marker area passed to ``Axes.scatter``.
- **`save_path`** — Destination passed to :func:`spacr.plot.save_figure`. The written extension follows the configured figure format.
- **`threshold`** — Absolute coefficient threshold for calls. A row is returned when ``p_value &lt;= 0.05`` and ``abs(coefficient) &gt;= abs(threshold)``.
- **`x_lim`** — Two x-axis limits. The default is ``[-0.5, 0.5]``.
- **`y_lims`** — Use ``[low, high]`` for one axis or ``[[lower_low, lower_high], [upper_low, upper_high]]`` for a broken y-axis. By default, fit one axis to the finite values.
