### [`spacr.toxo.custom_volcano_plot`](https://einarolafsson.github.io/spacr/api/spacr/toxo/index.html#spacr.toxo.custom_volcano_plot)


#### Plot

- **`x_lim`** *(optional)* — Two x-axis limits. The default is ``[-0.5, 0.5]``.
- **`y_lims`** *(optional)* — Use ``[low, high]`` for one axis or ``[[lower_low, lower_high], [upper_low, upper_high]]`` for a broken y-axis. By default, fit one axis to the finite values.

#### Additional settings

- **`data_path`** *(required)* — Regression table containing ``feature``, ``coefficient``, and ``p_value`` columns. DataFrame input is copied.
- **`draw`** *(optional)* — Build, optionally save, and show the figure. If false, return the hit list before constructing a figure.
- **`figsize`** *(optional)* — Width and height of the square figure in inches. Typography scales with this value.
- **`highlight_location`** *(optional)* — Values from ``metadata_column`` to overlay in the highlight colour and name in the in-panel legend.
- **`metadata_column`** *(optional)* — Localization or annotation column used by ``highlight_location``.
- **`metadata_path`** *(required)* — Gene metadata containing one row per ``gene_nr`` and the selected metadata column. DataFrame input is copied.
- **`point_size`** *(optional)* — Marker area passed to ``Axes.scatter``.
- **`save_path`** *(optional)* — Destination passed to :func:`spacr.figures.scene.write_figure`, which draws the scene the screen would show and falls back to :func:`spacr.plot.save_figure`. The written extension follows the configured figure format either way.
- **`threshold`** *(optional)* — Absolute coefficient threshold for calls. A row is returned when ``p_value &lt;= 0.05`` and ``abs(coefficient) &gt;= abs(threshold)``.
