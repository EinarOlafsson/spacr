```python
import inspect
from spacr.toxo import custom_volcano_plot as _function
settings = {key: None if parameter.default is inspect.Parameter.empty else parameter.default
            for key, parameter in inspect.signature(_function).parameters.items()
            if parameter.kind not in (inspect.Parameter.VAR_POSITIONAL, inspect.Parameter.VAR_KEYWORD)}
import copy
_defaults_settings = copy.deepcopy(settings)

settings.update({
    # Additional settings
    # Required settings
    'data_path': None,
    'metadata_path': None,
})
```

```python
advanced_overrides = {}
settings.update(advanced_overrides)
```

```python
try:
    from _spacr_notebook_widgets import show_advanced_settings
except ImportError:
    show_advanced_settings = None
    print('Optional editor helper unavailable; use the advanced reference and advanced_overrides.')
settings_editor = (show_advanced_settings(settings, ['data_path', 'metadata_path'], _defaults_settings)
                   if show_advanced_settings is not None else None)
try:
    from _spacr_notebook_widgets import show_help_menu
except ImportError:
    pass
else:
    help_menu = show_help_menu('reference/30_volcano_plot.md')
```

```python
custom_volcano_plot(**settings)
```
