**When to use it.** When a model trained on one screen should seed a related one rather than starting from scratch.

## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.deep_spacr.model_knowledge_transfer`](https://einarolafsson.github.io/spacr/api/spacr/deep_spacr/index.html#spacr.deep_spacr.model_knowledge_transfer)

- **`alpha`** — Weight on the true-label loss vs. distillation loss. Default ``0.5``.
- **`data_loader`** — Training DataLoader used during distillation.
- **`device`** — Torch device string. Default ``'cpu'``.
- **`dropout_rate`** — Optional dropout rate for the student.
- **`epochs`** — Training epochs. Default ``10``.
- **`lr`** — Adam learning rate. Default ``1e-4``.
- **`pretrained`** — Whether the student uses pretrained weights.
- **`student_model_name`** — TorchModel architecture name for the student. Default ``'maxvit_t'``.
- **`student_save_path`** — Destination for the trained student; the suffix ``_KD.pth`` is appended.
- **`teacher_paths`** — Paths to the teacher checkpoints (each either a saved ``TorchModel`` or a state dict).
- **`temperature`** — Softmax temperature for distillation. Default ``2.0``.
- **`use_checkpoint`** — Whether to enable gradient checkpointing.

## 4. Run

After editing the settings cell, run the function cell immediately below it. Long operations report progress through spaCR's logging system; set `SPACR_LOG_LEVEL=DEBUG` before starting Jupyter for more detail.

````python
settings = {
    'alpha': 0.5,
    'data_loader': None,
    'device': 'cpu',
    'dropout_rate': None,
    'epochs': 10,
    'lr': 0.0001,
    'pretrained': True,
    'student_model_name': 'maxvit_t',
    'student_save_path': None,
    'teacher_paths': None,
    'temperature': 2.0,
    'use_checkpoint': False,
}
````

````python
model_knowledge_transfer(**settings)
````
