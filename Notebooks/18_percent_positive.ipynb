### Every setting this function accepts

Every key, with its default and what it controls. Edit values in place; delete nothing — a key left at its default behaves exactly as if it were absent.

Generated from this installed version by `tools/build_notebook_settings.py`, so these are the real keys, the real defaults and the real descriptions. Re-run that tool after upgrading spaCR.

````python
# Generated by tools/build_notebook_settings.py — do not edit by hand.
# Values are this installed version's defaults; the text above each key is
# its live description. Edit values in place -- a key left at its default
# behaves exactly as if it were absent.

# spacr.submodules.analyze_percent_positive  (5 settings)
settings = {
    # (list or None) - Two-element [column, minimum] applied before
    # anything is counted: rows whose column value is not strictly greater
    # than minimum are dropped. Use it to exclude objects too small or too
    # dim to score, so they neither raise nor lower the percentage. It is
    # a strict > and takes ONE bound, so there is no upper limit and a
    # runaway-large object is kept. Default None - nothing is pre-
    # filtered.
    'filter_1': ['cell_area', 1000],

    # (str, path) - Folder the current step reads from and writes into:
    # raw images for mask generation, the merged/ folder of .npy stacks
    # for measure, the plate root for dataset/regression steps, or the
    # folder of .fastq.gz reads for sequencing. Outputs (stack/, masks/,
    # measurements/measurements.db, datasets/, results/) are created
    # inside it. A list of paths, or a "['a','b']" string, processes
    # several plates in one run. No usable default: the settings factories
    # fill a placeholder ('path' or '/path/to/src'), so this must be
    # supplied.
    'src': 'path',

    # (list) - Measurement tables read from each plate's database and
    # merged into one analysis frame. Only 'cell', 'nucleus', 'pathogen',
    # 'cytoplasm' and 'png_list' are actually merged. Any other name --
    # INCLUDING 'organelle', which the measure step does write -- is
    # loaded and then dropped, so asking for it costs time and returns
    # nothing, with no warning that the table you wanted is missing from
    # the result. Default ['cell', 'nucleus', 'pathogen', 'cytoplasm'].
    'tables': ['cell'],

    # (list, list-of-lists, int or None) - Cut-off applied to
    # 'measurement' before the annotation grid loads, so you only label
    # the objects you care about. Accepts a number or a quantile code
    # 'q1'-'q9' (q3 = the 30th percentile of that column), or one entry
    # per measurement when measurement is a list. Empty or None loads
    # every object unfiltered. Default 2000 where a numeric cutoff is
    # used; empty where the setting is optional.
    'threshold': 2000,

    # (str) - Measurement column compared against threshold to call each
    # object positive. Objects strictly above are annotated 'above' and
    # the rest 'below', and the percentage reported per well is the share
    # of 'above'. Point it at the column the phenotype lives in - a
    # recruitment ratio, a mean intensity - not at a size. Changing it
    # changes what 'positive' means, so two runs with different value_col
    # are not comparable. Default None; must be supplied.
    'value_col': 'cell_channel_2_mean_intensity',
}
````

````python
analyze_percent_positive(settings)
````
