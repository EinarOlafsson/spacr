**What it does.** Join sequencing counts against imaging scores and check they agree.

**When to use it.** A QC step between barcode mapping and regression — catches plate mix-ups and swapped wells before they become a result.

**What you get.** Correlation plots and a per-well joined table.

## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.submodules.compare_reads_to_scores`](https://einarolafsson.github.io/spacr/api/spacr/submodules/index.html#spacr.submodules.compare_reads_to_scores)

- **`column`** — column used to select a subset of wells. Default ``'columnID'``.
- **`empirical_dict`** — mapping of ``rowID`` to ``(pc_units, nc_units)`` mixture; a 16-row default is used when ``None``.
- **`nc_grna`** — negative-control gRNA name. Default ``'TGGT1_233460_4'``.
- **`pc_grna`** — positive-control gRNA name. Default ``'TGGT1_220950_1'``.
- **`plate`** — plate ID to stamp when a single pair of CSVs is given.
- **`reads_csv`** — path (or list of paths) to per-gRNA read count CSVs.
- **`save_paths`** — two-element list of PDF output paths (pc plot, nc plot).
- **`scores_csv`** — path (or list of paths) to per-object classifier score CSVs.
- **`value`** — value in ``column`` to keep. Default ``'c3'``.
- **`y_columns`** — columns to plot on the y axis; a sensible default is used when ``None``.

## 4. Run

After editing the settings cell, run the function cell immediately below it. Long operations report progress through spaCR's logging system; set `SPACR_LOG_LEVEL=DEBUG` before starting Jupyter for more detail.

````python
settings = {
    'column': 'columnID',
    'empirical_dict': None,
    'nc_grna': 'TGGT1_233460_4',
    'pc_grna': 'TGGT1_220950_1',
    'plate': None,
    'reads_csv': None,
    'save_paths': None,
    'scores_csv': None,
    'value': 'c3',
    'y_columns': None,
}
````

````python
compare_reads_to_scores(**settings)
````
