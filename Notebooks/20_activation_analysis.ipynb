- **`model`** *(required)* — Trained classifier.
- **`n_steps`** *(optional)* — Number of perturbation steps for deletion and insertion curves.
- **`sanity_threshold`** *(optional)* — Maximum rank correlation at which a randomized map is considered sufficiently different.
