## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.deep_spacr.analyze_activation_maps`](https://einarolafsson.github.io/spacr/api/spacr/deep_spacr/index.html#spacr.deep_spacr.analyze_activation_maps)

- **`baseline`** — what removed pixels become — ``'blur'``, ``'zero'``, ``'mean'`` or ``'uniform'``.
- **`images`** — one image tensor, or a sequence of them, ``(C, H, W)``.
- **`masks`** — optional per-image boolean object masks for the pointing game.
- **`methods`** — method names from :data:`spacr.attribution.ATTRIBUTION_METHODS`; defaults to one representative of each family.
- **`model`** — the trained classifier.
- **`model_type`** — architecture name, used to make errors readable.
- **`n_steps`** — steps in the deletion / insertion curves.
- **`sanity_check`** — run the randomisation check (on the first image).
- **`sanity_threshold`** — rank correlation below which a method passes.
- **`target`** — class index to explain; defaults to each image's prediction.
- **`target_layer`** — CAM target layer, or None for the last convolution.
- **`verbose`** — print the per-method verdicts.

````python
analyze_activation_maps(**settings)
````
