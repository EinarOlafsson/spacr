### [`spacr.submodules.generate_score_heatmap`](https://einarolafsson.github.io/spacr/api/spacr/submodules/index.html#spacr.submodules.generate_score_heatmap)

> This function does not yet expose a settings factory. The list below is recovered from direct settings access in its source, so a key used only on a dynamic branch may be absent.


#### Plate Layout & Controls

- **`plateID`** *(optional)* — (str) - Plate name stamped onto count and score rows that carry no plate of their own, and used as the first field of the plate_row_column key that joins the two tables. It is ignored with a warning when the input already contains more than one distinct plate, so it matters only for single-plate inputs. Default 'plate1'.

#### Plot

- **`cmap`** *(optional)* — (str) - Matplotlib colormap applied to single-channel image previews and to plate heatmaps. Perceptually uniform maps ('viridis', 'inferno', 'magma') keep intensity differences honest; 'gray' matches how the raw microscope data looks. Any registered matplotlib name works, with an '_r' suffix to reverse it. Default 'inferno' for image plots, 'viridis' for plate heatmaps.

#### Additional settings

- **`columnID`** *(optional)* — (str) - Plate-map column identifying the plate column - the 1-24 axis of a well name - used to group wells and to lay the heatmap out along its x axis. Naming a column that is not the plate column produces a heatmap whose axis is meaningless rather than an error. Default None.
- **`control_sgrnas`** *(optional)* — (list) - gRNA names treated as controls when the mixed-condition fraction is computed; every fraction is measured relative to these. A wrong or incomplete list shifts every fraction on the plate in the same direction rather than raising, so check it against the library. Default None.
- **`csv`** *(optional)* — (str, path) - Sequencing-derived CSV of gRNA counts per well, used to compute the mixed-condition fraction the scores are compared against. Distinct from csv_name, which names the per-plate score files; this one is a single file for the whole comparison. Default None.
- **`csv_name`** *(optional)* — (str) - Filename searched for inside every entry of folders. All plates must use the same name - a plate whose file is named differently is skipped silently and its column is simply missing from the heatmap rather than reported. Default None.
- **`cv_csv`** *(optional)* — (str, path) - CSV of cross-validated scores, added to the heatmap as its own row so a held-out score can be read directly beside the in-sample ones. Leave it None to plot the score rows alone. Default None.
- **`data_column`** *(optional)* — (str) - Column in each per-plate score CSV holding the value to plot. It must exist in every file named by csv_name; a plate missing it drops out of the heatmap. This is the number the whole figure shows, so naming the wrong column produces a plausible heatmap of something else. Default None.
- **`data_column_cv`** *(optional)* — (str) - Column in cv_csv holding the cross-validated value. Named separately from data_column because a cross-validation file almost always labels its column differently from the score files it came from. Default None.
- **`dst`** *(optional)* — (str) - Folder receiving versioned tables, manifests and figures. Default '' uses a module-specific folder beside the primary input, keeping different analyses separated.
- **`folders`** *(optional)* — (list) - One folder per plate holding the classification CSVs to combine into a single heatmap; each is searched for csv_name. The order given is the order the plates are laid out in. Default None.
- **`fraction_grna`** *(optional)* — (str) - The single gRNA whose fraction is kept and plotted, selected out of the fraction table once it has been computed. Everything else in that table is discarded, so this picks which gRNA the comparison is about. Default None.
