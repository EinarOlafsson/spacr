# Score heatmap

**What it does.** Lay per-well scores out in plate geometry.

**When to use it.** To see edge effects, gradients and dead corners — plate position artefacts are invisible in a scatter plot and obvious here.

**What you get.** A plate-shaped heatmap per score column.

### Every setting this function accepts

Every key, with its default and what it controls. Edit values in place; delete nothing — a key left at its default behaves exactly as if it were absent.

Generated from this installed version by `tools/build_notebook_settings.py`, so these are the real keys, the real defaults and the real descriptions. Re-run that tool after upgrading spaCR.

````python
# Generated by tools/build_notebook_settings.py — do not edit by hand.
# Values are this installed version's defaults; the text above each key is
# its live description. Edit values in place -- a key left at its default
# behaves exactly as if it were absent.

# spacr.submodules.generate_score_heatmap  (12 settings)
# NOTE: this function has no settings helper declaring its keys -- the list
# below was read from its source, so a key it reads only on one branch may
# be missing. See instruction 88.
settings = {
    # (str) - Matplotlib colormap applied to single-channel image previews
    # and to plate heatmaps. Perceptually uniform maps ('viridis',
    # 'inferno', 'magma') keep intensity differences honest; 'gray'
    # matches how the raw microscope data looks. Any registered matplotlib
    # name works, with an '_r' suffix to reverse it. Default 'inferno' for
    # image plots, 'viridis' for plate heatmaps.
    'cmap': None,

    # (str) - Plate-map column identifying the plate column - the 1-24
    # axis of a well name - used to group wells and to lay the heatmap out
    # along its x axis. Naming a column that is not the plate column
    # produces a heatmap whose axis is meaningless rather than an error.
    # Default None.
    'columnID': None,

    # (list) - gRNA names treated as controls when the mixed-condition
    # fraction is computed; every fraction is measured relative to these.
    # A wrong or incomplete list shifts every fraction on the plate in the
    # same direction rather than raising, so check it against the library.
    # Default None.
    'control_sgrnas': None,

    # (str, path) - Sequencing-derived CSV of gRNA counts per well, used
    # to compute the mixed-condition fraction the scores are compared
    # against. Distinct from csv_name, which names the per-plate score
    # files; this one is a single file for the whole comparison. Default
    # None.
    'csv': None,

    # (str) - Filename searched for inside every entry of folders. All
    # plates must use the same name - a plate whose file is named
    # differently is skipped silently and its column is simply missing
    # from the heatmap rather than reported. Default None.
    'csv_name': None,

    # (str, path) - CSV of cross-validated scores, added to the heatmap as
    # its own row so a held-out score can be read directly beside the in-
    # sample ones. Leave it None to plot the score rows alone. Default
    # None.
    'cv_csv': None,

    # (str) - Column in each per-plate score CSV holding the value to
    # plot. It must exist in every file named by csv_name; a plate missing
    # it drops out of the heatmap. This is the number the whole figure
    # shows, so naming the wrong column produces a plausible heatmap of
    # something else. Default None.
    'data_column': None,

    # (str) - Column in cv_csv holding the cross-validated value. Named
    # separately from data_column because a cross-validation file almost
    # always labels its column differently from the score files it came
    # from. Default None.
    'data_column_cv': None,

    # (str) - Folder for the QC figures, tables and written
    # recommendation. Empty writes a 'barcode_qc' folder beside the first
    # count_data file. Default ''.
    'dst': None,

    # (list) - One folder per plate holding the classification CSVs to
    # combine into a single heatmap; each is searched for csv_name. The
    # order given is the order the plates are laid out in. Default None.
    'folders': None,

    # (str) - The single gRNA whose fraction is kept and plotted, selected
    # out of the fraction table once it has been computed. Everything else
    # in that table is discarded, so this picks which gRNA the comparison
    # is about. Default None.
    'fraction_grna': None,

    # (str) - Plate name stamped onto count and score rows that carry no
    # plate of their own, and used as the first field of the
    # plate_row_column key that joins the two tables. It is ignored with a
    # warning when the input already contains more than one distinct
    # plate, so it matters only for single-plate inputs. Default 'plate1'.
    'plateID': None,
}
````

````python
generate_score_heatmap(settings)
````
