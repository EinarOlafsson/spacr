# Fuse several models

**What it does.** Combine several trained classifiers into one predictor.

**When to use it.** When individual models each capture part of the phenotype and their errors are not the same.

**What you get.** A fused model and its combined performance report.

## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.deep_spacr.model_fusion`](https://einarolafsson.github.io/spacr/api/spacr/deep_spacr/index.html#spacr.deep_spacr.model_fusion)

- **`aggregator`** — Reduction over stacked weights — one of ``'mean'``, ``'geomean'``, ``'median'``, ``'sum'``, ``'max'``, ``'min'``. Default ``'mean'``.
- **`device`** — Torch device string. Default ``'cpu'``.
- **`dropout_rate`** — Optional dropout rate.
- **`model_name`** — TorchModel architecture name for the fused model.
- **`model_paths`** — Paths to source checkpoints (dicts or ``TorchModel`` objects).
- **`pretrained`** — Whether pretrained weights are expected.
- **`save_path`** — Base output path; suffix ``_&lt;aggregator&gt;.pth`` is appended.
- **`use_checkpoint`** — Whether to enable gradient checkpointing.

## 4. Run

After editing the settings cell, run the function cell immediately below it. Long operations report progress through spaCR's logging system; set `SPACR_LOG_LEVEL=DEBUG` before starting Jupyter for more detail.

````python
settings = {
    'aggregator': 'mean',
    'device': 'cpu',
    'dropout_rate': None,
    'model_name': 'maxvit_t',
    'model_paths': None,
    'pretrained': True,
    'save_path': None,
    'use_checkpoint': False,
}
````

````python
model_fusion(**settings)
````
