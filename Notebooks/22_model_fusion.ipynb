### [`spacr.deep_spacr.model_fusion`](https://einarolafsson.github.io/spacr/api/spacr/deep_spacr/index.html#spacr.deep_spacr.model_fusion)


#### Computer Vision Model

- **`model_name`** *(optional)* — TorchModel architecture name for the fused model.

#### Computer Vision Optimization and Regularization

- **`use_checkpoint`** *(optional)* — Whether to enable gradient checkpointing.
- **`dropout_rate`** *(optional)* — Optional dropout rate.

#### Additional settings

- **`aggregator`** *(optional)* — Reduction over stacked weights — one of ``'mean'``, ``'geomean'``, ``'median'``, ``'sum'``, ``'max'``, ``'min'``. Default ``'mean'``.
- **`device`** *(optional)* — Torch device string. Default ``'cpu'``.
- **`model_paths`** *(required)* — Paths to source checkpoints (dicts or ``TorchModel`` objects).
- **`pretrained`** *(optional)* — Whether pretrained weights are expected.
- **`save_path`** *(required)* — Base output path; suffix ``_&lt;aggregator&gt;.pth`` is appended.
