## 3. Settings and API reference

Read the descriptions here, then edit only the values in the next cell. Defaults and descriptions are generated from the installed spaCR version, so the notebook stays aligned with the API.

### [`spacr.toxo.plot_gene_phenotypes`](https://einarolafsson.github.io/spacr/api/spacr/toxo/index.html#spacr.toxo.plot_gene_phenotypes)

- **`data`** — Gene identifiers and phenotype/error columns. The frame is copied before numeric conversion.
- **`data_column`** — Mean phenotype column plotted on the y-axis.
- **`error_column`** — Standard-error column used for the uncertainty band.
- **`gene_list`** — Gene names or ``TGGT1_&lt;id&gt;`` identifiers to highlight.
- **`save_path`** — Figure destination. The configured figure format controls the final extension.
- **`x_column`** — Column used to match gene identifiers.

## 4. Run

````python
settings = {
    'data': None,
    'data_column': 'T.gondii GT1 CRISPR Phenotype - Mean Phenotype',
    'error_column': 'T.gondii GT1 CRISPR Phenotype - Standard Error',
    'gene_list': None,
    'save_path': None,
    'x_column': 'Gene ID',
}
````

````python
plot_gene_phenotypes(**settings)
````
